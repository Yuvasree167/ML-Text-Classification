```python
import joblib

from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, classification_report
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline
```

```python
print("Loading 20 Newsgroups dataset...")

train = fetch_20newsgroups(
    subset='train',
    remove=('headers', 'footers', 'quotes')
)

test = fetch_20newsgroups(
    subset='test',
    remove=('headers', 'footers', 'quotes')
)

print("Training samples:", len(train.data))
print("Testing samples:", len(test.data))
print("Number of categories:", len(train.target_names))
```

```python
print("Building the Multinomial Naive Bayes model pipeline...")

model = make_pipeline(
    TfidfVectorizer(),
    MultinomialNB()
)

print("Model created successfully!")
```

```python
print("Training the model...")

model.fit(train.data, train.target)

print("Training completed!")
```

```python
predicted = model.predict(test.data)

accuracy = accuracy_score(test.target, predicted)

print(f"Test Accuracy: {accuracy * 100:.2f}%")
```

```python
print("\nClassification Report:\n")

print(
    classification_report(
        test.target,
        predicted,
        target_names=test.target_names
    )
)
```

```python
model_filename = "20newsgroups_model.joblib"

joblib.dump(model, model_filename)

print(f"Model saved successfully as {model_filename}")
```

```python
loaded_model = joblib.load("20newsgroups_model.joblib")

new_samples = [
    "The local team won the championship game last night.",
    "Scientists discovered a new planet using a telescope.",
    "My computer graphics card is overheating."
]

predictions = loaded_model.predict(new_samples)

for text, prediction in zip(new_samples, predictions):
    print("Text:", text)
    print("Predicted Category:", train.target_names[prediction])
    print()
```

```python
%%writefile train.py

import joblib

from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, classification_report
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline

print("Loading 20 Newsgroups dataset...")

train = fetch_20newsgroups(
    subset='train',
    remove=('headers', 'footers', 'quotes')
)

test = fetch_20newsgroups(
    subset='test',
    remove=('headers', 'footers', 'quotes')
)

print("Building the Multinomial Naive Bayes model pipeline...")

model = make_pipeline(
    TfidfVectorizer(),
    MultinomialNB()
)

print("Training the model...")

model.fit(train.data, train.target)

predicted = model.predict(test.data)

accuracy = accuracy_score(test.target, predicted)

print(f"Test Accuracy: {accuracy * 100:.2f}%")

print("\nClassification Report:")

print(
    classification_report(
        test.target,
        predicted,
        target_names=test.target_names
    )
)

model_filename = "20newsgroups_model.joblib"

joblib.dump(model, model_filename)

print(f"Model saved successfully as {model_filename}")
```
